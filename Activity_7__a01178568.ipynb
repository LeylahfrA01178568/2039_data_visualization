```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
path = '/content/antropometria-dataset.csv'
df = pd.read_csv(path)
```

```python
df.head()
```

```python
columns_act = ['sistol','sexo', 'diastol', 'peso', 'cintura', 'cadera', 'edad']
ndf = df[columns_act].copy()
ndf
```

```python
ndf.head()
```

## Histograms of  blood pressure variables 'systol' and 'diastol'

```python
plt.hist(df['sistol'], bins=15)
plt.title('Systolic Blood Pressure')
plt.xlabel('Systolic Pressure')
plt.ylabel('Frequency')
plt.show()

plt.hist(df['diastol'], bins=15)
plt.title('Diastolic Blood Pressure')
plt.xlabel('Diastolic Pressure')
plt.ylabel('Frequency')
plt.show()
```

### By sex


```python

sex_1 = df[df['sexo'] == 1]
sex_2 = df[df['sexo'] == 2]
```

```python

plt.figure(figsize=(12,5))

plt.subplot(1,2,1)
plt.hist(sex_1['sistol'], bins=30)
plt.title('Systolic Blood Pressure - Men')
plt.xlabel('Systolic Pressure')
plt.ylabel('Frequency')

plt.subplot(1,2,2)
plt.hist(sex_2['sistol'], bins=30)
plt.title('Systolic Blood Pressure - Women')
plt.xlabel('Systolic Pressure')
plt.ylabel('Frequency')

plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(12,5))

plt.subplot(1,2,1)
plt.hist(sex_1['diastol'], bins=30)
plt.title('Diastolic Blood Pressure - Men')
plt.xlabel('Diastolic Pressure')
plt.ylabel('Frequency')

plt.subplot(1,2,2)
plt.hist(sex_2['diastol'], bins=30)
plt.title('Diastolic Blood Pressure - Women')
plt.xlabel('Diastolic Pressure')
plt.ylabel('Frequency')

plt.tight_layout()
plt.show()
```

## Histograms of 'peso' and 'cintura'

```python

plt.hist(df['peso'], bins=30)
plt.title('Weight Distribution')
plt.xlabel('Weight (kg)')
plt.ylabel('Frequency')
plt.show()

plt.hist(df['cintura'], bins=30)
plt.title('Waist Distribution')
plt.xlabel('Waist (cm)')
plt.ylabel('Frequency')
plt.show()
```

```python
plt.figure(figsize=(12,5))

plt.subplot(1,2,1)
plt.hist(sex_1['peso'], bins=30)
plt.title('Weight Distribution - Men')
plt.xlabel('Weight (kg)')
plt.ylabel('Frequency')

plt.subplot(1,2,2)
plt.hist(sex_2['peso'], bins=30)
plt.title('Weight Distribution - Women')
plt.xlabel('Weight (kg)')
plt.ylabel('Frequency')
```

## Scatterplot to analyze the relationship between the variables 'peso' and 'cintura'.

```python
plt.scatter(df['peso'], df['cintura'], alpha=0.5)
plt.title('Weight vs Waist Circumference')
plt.xlabel('Weight (kg)')
plt.ylabel('Waist (cm)')
plt.show()
```

```python
plt.figure(figsize=(12,5))

plt.subplot(1,2,1)
plt.scatter(sex_1['peso'], sex_1['cintura'], alpha=0.5)
plt.title('Weight vs Waist - Men')
plt.xlabel('Weight (kg)')
plt.ylabel('Waist (cm)')

plt.subplot(1,2,2)
plt.scatter(sex_2['peso'], sex_2['cintura'], alpha=0.5)
plt.title('Weight vs Waist - Women')
plt.xlabel('Weight (kg)')
plt.ylabel('Waist (cm)')

plt.tight_layout()
plt.show()
```

### If 'sexo 2 = mujer', and 'sexo 1 = hombre', there are outliers in mujer_cintura

```python

women_waist = sex_2['cintura']

Q1 = women_waist.quantile(0.25)
Q3 = women_waist.quantile(0.75)
IQR = Q3 - Q1

lower_limit = Q1 - 1.5 * IQR
upper_limit = Q3 + 1.5 * IQR

outliers = (women_waist < lower_limit) | (women_waist > upper_limit)

print(outliers.any())
```

```python
plt.boxplot(women_waist.dropna())
plt.title('Women Waist Circumference')
plt.ylabel('Waist Circumference (cm)')
plt.show()
```
