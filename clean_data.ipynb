### 3. Drop all the rows, if all the values are missing

### 7. Replace missing values with descriptive statistics

For categorical values, when you want to replace the missing value, use mode "descriptive statistics"

Mean (average) is better when there are not outliers.

Median is better when there are outliers.

```python
df.describe ()
```

```python
df.Stars
```

```python
sm = df.Stars.mean()
df['Stars'].fillna(sm, inplace = True) # fill the missing values with stars mean
df.Stars
```

```python
wm = df.Weight.mean()
df['Weight'].fillna(sm, inplace = True) # fill the missing values with stars mean
df.Weight
```

### RQ1 :  what is the average Weigh of cars made in Mexico

```python
df
```

```python
origin_info = pd.pivot_table(
    df,
    index = 'Origin', # to know what goes in rows and columns
    aggfunc= 'mean',
    values = 'Weight'
)
origin_info
```

```python
origin_info.loc['Mexico']
print('Average Weight of cars made in Mexico is \t:', origin_info.loc['Mexico'])
```

### RQ.2 what is the Average Weight of Ford cars made in mexico?

```python
ford_info = pd.pivot_table(
    df,
    index=['Origin', 'Company'], # Group by both Origin and Company
    aggfunc= 'mean',
    values = 'Weight' # Only aggregate the numeric 'Weight' column
)
ford_info
```

```python
ford_info.loc['Mexico','Ford']
```

```python
ford_info_col = pd.pivot_table(
    df,
    index=['Origin'], # Group by both Origin and Company
    aggfunc= 'mean',
    values = 'Weight', # Only aggregate the numeric 'Weight' column
    columns = 'Company'
)
ford_info_col
```

```python
ford_info_col.loc['Mexico', 'Ford']
print('The average weight for Ford cars in Mexico \t',ford_info_col.loc['Mexico', 'Ford'] )
```
