```python
import pandas as pd
import numpy as np
```

```python
antropometría_path = 'https://raw.githubusercontent.com/tec03/Datasets/refs/heads/main/datasets/antropometria-dataset.csv'
```

```python

df_antro = pd.read_csv(antropometría_path)
df_antro.head(5)
```

```python
df_antro.info()
```

### 1. `Pivot table` showing average weight of people per state

```python
# average weight (peso) by state (desc_ent)
pivot_weight_by_state = pd.pivot_table(
    df_antro,
    values='peso',
    index='desc_ent',
    aggfunc='mean'
)


display(pivot_weight_by_state)
```

### `sex variable`

Understanding that the sexo variable is a categorical value, sex 1 and 2

```python

sex_count = df_antro['sexo'].value_counts()
display(sex_count) # it has descrete values, identify the type of values
```

```python

df_sex_weight_table = pd.pivot_table(
    df_antro,
    values='peso',
    index='desc_ent',
    columns='sexo',
    aggfunc='mean'
)

display(df_sex_weight_table)
```

#### Top 5 states with higher average weight per  gender



```python
# Used .nlargest() to obtain directly the highest values
print("Top 5 Estados - Sexo 1 (Hombres):")
display(df_sex_weight_table[1].nlargest(5))

print("\nTop 5 Estados - Sexo 2 (Mujeres):")
display(df_sex_weight_table[2].nlargest(5))
```

```python
# Used .nsmallest() to obtain directly the lowest values
print("Top 5 Estados - Sexo 1 (Hombres):")
display(df_sex_weight_table[1].nsmallest(5))

print("\nTop 5 Estados - Sexo 2 (Mujeres):")
display(df_sex_weight_table[2].nsmallest(5))
```

```python

print('El estado con el peso promedio más bajo en sexo 1 es:\n', df_sex_weight_table[1].nsmallest().index[0])
```

```python
display(df_sex_weight_table[1].nsmallest(1))
```

### 2. `Pivot table`  number of people by each marital status.

```python
adultos_path ='https://raw.githubusercontent.com/tec03/Datasets/refs/heads/main/datasets/adult.csv'
```

```python
df_adult = pd.read_csv(adultos_path)
df_adult.head(5)
```

## Create a pivot table by cpunting de number of ages registered, adn then divide per marital status

```python

pivot_marital_status = pd.pivot_table(
    df_adult,
    values='age',
    index='marital-status',
    aggfunc='count'
).rename(columns={'age': 'Number of People'})

display(pivot_marital_status)
```

### Percentage column


Add a column showing the percentage of people by each marital status.

```python
pivot_marital_status = pd.pivot_table(
    df_adult,
    values='age',
    index='marital-status',
    aggfunc='count'
).rename(columns={'age': 'Number of People'})
pivot_marital_status
```

```python
percent= (pivot_marital_status['Number of People'] / pivot_marital_status['Number of People'].sum()) * 100
```

```python

pivot_marital_status['Percentage (%)'] = percent

display(pivot_marital_status)
```

#### What is the average age of Divorced with HS-grad?



```python
marital_stat_education = pd.pivot_table(
    df_adult,
    values='age',
    index=['education', 'marital-status'],
    aggfunc='mean'
)
marital_stat_education
```

####  `'median' 'hours-per-week' of 'Male' 'Married-AF-spouse' with education 'Bachelors'`

```python
pivot_marital_status_sex_hours = pd.pivot_table(
    df_adult,
    values=['age', 'hours-per-week'],
    index=['marital-status','education'],
    columns='sex',
    aggfunc='median'
)
pivot_marital_status_sex_hours
```

```python

quiz_answer = pivot_marital_status_sex_hours.loc[('Married-AF-spouse', 'Bachelors'), ('hours-per-week','Male')]
quiz_answer
```
