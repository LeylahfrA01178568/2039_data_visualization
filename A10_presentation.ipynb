# [5. Q3: How does physical activity frequency differ between individuals with obesity and other weight classifications?](#scrollTo=u9q-QGBEvqn_&line=11&uniqifier=1) **histogram**

**Study:** Analyze how physical activity frequency is distributed between individuals with obesity and other weight classifications to identify differences in concentration, peaks, and variability.

**Why a Histogram?:** A histogram allows us to compare the distributions of a numerical variable, making it easier to identify which physical activity levels are most concentrated within each group.

```python

# Separate observations into two groups
obesity_group = ndf[ndf['has_obesity'] == True]
other_group = ndf[ndf['has_obesity'] == False]
```

## 5.1 Figure 3. Histogram

```python


from matplotlib.patches import Patch

# Define the bins and calculate histogram densities
bins = np.linspace(0, 3, 16)

obesity_density, edges = np.histogram(
    obesity_group['Physical_Act'],
    bins=bins,
    density=True
)

other_density, edges = np.histogram(
    other_group['Physical_Act'],
    bins=bins,
    density=True
)

# Find the overlapping density
overlap = np.minimum(obesity_density, other_density)

# Define bar positions
positions = edges[:-1]
widths = np.diff(edges)

fig, ax = plt.subplots(figsize=(15, 6))

# Plot both distributions
ax.bar(
    positions, obesity_density,
    width=widths, align='edge',
    color='lightcoral',
    edgecolor='black',
    alpha=0.8
)

ax.bar(
    positions, other_density,
    width=widths, align='edge',
    color='lightskyblue',
    edgecolor='black',
    alpha=0.8
)

# Add explicit purple overlap
ax.bar(
    positions, overlap,
    width=widths, align='edge',
    color='mediumpurple',
    edgecolor='black',
    alpha=0.9
)

# Add shadows
for bar in ax.patches:
    bar.set_path_effects([
        pe.SimplePatchShadow(offset=(3, -3), alpha=0.2),
        pe.Normal()
    ])

# Add numerical density labels
for i in range(len(positions)):
    maximum = max(obesity_density[i], other_density[i])

    if maximum > 0.1:
        ax.text(
            positions[i] + widths[i] / 2,
            maximum + 0.035,
            f'{maximum:.2f}',
            ha='center',
            fontsize=8
        )

# Set background
ax.set_facecolor('#F2F2F2')

# Legend
legend_elements = [
    Patch(facecolor='lightcoral', label='Obesity'),
    Patch(facecolor='lightskyblue', label='Other Classifications'),
    Patch(facecolor='mediumpurple', label='Overlap')
]

ax.legend(handles=legend_elements)

# Titles and labels
plt.title('Physical Activity Distribution by Obesity Classification')
plt.xlabel('Physical Activity Frequency (0-3)')
plt.ylabel('Density')

plt.ylim(0, max(obesity_density.max(), other_density.max()) * 1.15)
plt.tight_layout()
plt.show()
```

**Color purple  means that both groups have observations within the same physical activity frequency interval.**

## 5.2 Insight

1. Both groups show an irregular distribution, with the highest concentration of individuals at a physical activity frequency of 0.

2. Individuals in other weight classifications show higher concentrations around activity levels 1, 2, and 3, while those classified with obesity are more concentrated at lower and intermediate activity levels.

## 7.1 Pivot table

```python
# Count individuals by gender and weight classification

gender_pivot = pd.pivot_table(
    ndf,
    index='Gender_binary',
    columns='Weight_class',
    aggfunc='size',
    fill_value=0
)

# Calculate percentages within each gender
gender_pct = gender_pivot.div(
    gender_pivot.sum(axis=1),
    axis=0
) * 100

# Rename gender categories
gender_pct.index = gender_pct.index.map({
    1: 'Female',
    2: 'Male'
})

display(gender_pct.round(1))
```

# [9. Q7:How does the distribution of weight classifications differ depending on frequent high-calorie food consumption?](#scrollTo=u9q-QGBEvqn_&line=14&uniqifier=1) **Dumbbell plot**


**Study:** Calculate the percentage of individuals in each weight classification depending on whether they frequently consume high-calorie foods, to identify differences in their distribution patterns.

**Why a Dumbbell Plot?:** A dumbbell plot allows us to directly compare the percentages of both high-calorie consumption groups within each weight classification. The distance between the dots makes their differences easier to identify, without suggesting a continuous relationship between categorical variables.

## 9.1 Pivot table

```python

# Calculate percentages by weight classification

calorie_pivot = pd.pivot_table(
    ndf,
    index='Weight_class',
    columns='High_calorie',
    aggfunc='size',
    fill_value=0
)

calorie_pct = calorie_pivot.div(
    calorie_pivot.sum(axis=0),
    axis=1
) * 100

weight_order = [
    'Insufficient_Weight',
    'Normal_Weight',
    'Overweight_Level_I',
    'Overweight_Level_II',
    'Obesity_Type_I',
    'Obesity_Type_II',
    'Obesity_Type_III'
]

calorie_pct = calorie_pct.loc[weight_order]

display(calorie_pct.round(1))
```

## 9.2 Figure 7. Dumbbell Plot

```python
import matplotlib.pyplot as plt
import matplotlib.patheffects as pe
import numpy as np

# Position of weight categories
y = np.arange(len(calorie_pct))

fig, ax = plt.subplots(figsize=(13, 7))

# Connect the two percentages in each category
for i in range(len(calorie_pct)):
    ax.plot(
        [calorie_pct['no'].iloc[i], calorie_pct['yes'].iloc[i]],
        [y[i], y[i]],
        color='darkgrey',
        linewidth=3,
        zorder=1
    )

# Plot the percentages for both groups
ax.scatter(
    calorie_pct['no'], y,
    color='lightskyblue',
    edgecolor='steelblue',
    s=130,
    label='No',
    zorder=3
)

ax.scatter(
    calorie_pct['yes'], y,
    color='lightcoral',
    edgecolor='firebrick',
    s=130,
    label='Yes',
    zorder=3
)

# Add percentage labels
for i in range(len(calorie_pct)):
    ax.text(calorie_pct['no'].iloc[i],
            y[i] - 0.22,
            f"{calorie_pct['no'].iloc[i]:.1f}%",
            ha='center', fontsize=9)

    ax.text(calorie_pct['yes'].iloc[i],
            y[i] + 0.28,
            f"{calorie_pct['yes'].iloc[i]:.1f}%",
            ha='center', fontsize=9)

# Formatting
ax.set_yticks(y)
ax.set_yticklabels([
    'Underweight', 'Normal Weight', 'Overweight I',
    'Overweight II', 'Obesity I', 'Obesity II', 'Obesity III'
])

ax.invert_yaxis()
ax.set_xlim(0, 40)
ax.set_facecolor('#F2F2F2')
ax.grid(axis='x', color='lightblue', linestyle='--', alpha=0.4)
ax.set_axisbelow(True)

plt.title('Weight Classification by High-Calorie Food Consumption')
plt.xlabel(' Percentage Within Consumption Group (%)')
plt.ylabel('Weight Classification')
plt.legend(title='Frequent High-Calorie Food Consumption')

plt.tight_layout()
plt.show()
```

## 9.3 Insights

1. Normal Weight was more common among individuals who do not frequently consume high-calorie foods (32.1% vs. 11.1%), a difference of 21 percentage points.

2. Frequent high-calorie food consumers had higher proportions across all obesity classifications, with the greatest difference in Obesity Type III (17.5% vs. 0.4%).

# [10. Q8:To what extent does physical activity frequency explain variations in daily water intake](#scrollTo=u9q-QGBEvqn_&line=12&uniqifier=1)**Linear regression**

**Study:** Fit an Ordinary Least Squares (OLS) regression model to evaluate the strength, direction, and statistical significance of the relationship between physical activity frequency and daily water intake.

**Why a Regression Plot:** A regression plot allows us to compare observed water intake values with the fitted regression line.

## 10.1 Pivot table

```python


water_intake = pd.pivot_table(
    ndf,
    index=ndf.index,
    values=['Physical_Act', 'Water_intake']
)

display(water_intake.iloc[:5])
```

## 10.2 Figure 8. Linear regression model

```python
import statsmodels.formula.api as smf

# Fit the OLS regression model
model_water = smf.ols(
    formula='Water_intake ~ Physical_Act',
    data=water_intake
).fit()

# Display the regression summary
model_water.summary()
```

```python
print('Intercept:', round(model_water.params['Intercept'], 3))
print('Physical activity coefficient:', round(model_water.params['Physical_Act'], 3))
print('R-squared:', round(model_water.rsquared, 3))
print('P-value:', round(model_water.pvalues['Physical_Act'], 4))
```

```python
from IPython.display import display, Math

# Obtain regression coefficients
b0 = model_water.params['Intercept']
b1 = model_water.params['Physical_Act']

# Display the regression equation
display(Math(
    r'\text{Water\ intake} = ' + f'{b0:.3f} + {b1:.3f}' + r'(\text{Physical\_Act})'
))
```

```python
# Create results DataFrame
water_results = water_intake.copy()

# Add predicted values and residuals from OLS model
water_results['Predicted_Water'] = model_water.fittedvalues
water_results['Residual'] = model_water.resid

display(water_results.head(10))
```
